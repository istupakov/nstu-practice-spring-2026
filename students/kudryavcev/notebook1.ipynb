```python
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)
plt.title("График функции sin(x)")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.grid()
plt.show()
```

```python
plt.hist(df["Баллы"].dropna(), bins=5)
plt.title("Распределение баллов")
plt.xlabel("Баллы")
plt.ylabel("Количество")
plt.show()
```

```python
group_mean = df.groupby("Группа")["Баллы"].mean()

plt.bar(group_mean.index, group_mean.values)
plt.title("Средний балл по группам")
plt.xlabel("Группа")
plt.ylabel("Средний балл")
plt.show()
```

```python
plt.scatter(df["Возраст"], df["Баллы"])
plt.title("Возраст vs Баллы")
plt.xlabel("Возраст")
plt.ylabel("Баллы")
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Возраст"], df["Баллы"])
plt.title("Возраст vs Баллы")
plt.xlabel("Возраст")
plt.ylabel("Баллы")
plt.grid(True)

plt.show()
```
